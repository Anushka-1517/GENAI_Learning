```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

```python
### Discuss the Problem First.
```

```python
text = "Hello, My name is Anushka Sangal, " \
"Email is anushkasangal15@gmail.com and my age is 20"
```

```python
res = llm.invoke(f"Please give me only name, email and age from this text: {text}")
```

```python
print(res.content)
```

```python
### Solution is: Structured Output
```

```python
from pydantic import BaseModel, Field
from typing import List
class ResponseStructure(BaseModel):
    name:str = Field(description="Full Name ")
    email:str = Field(description="Email Address")
    age:int = Field(description="Age")

structured_llm = llm.with_structured_output(ResponseStructure)
```

```python
res = structured_llm.invoke(f"Please give me only name, email and age from this text: {text}")
```

```python
res.model_dump()
```

```python
from pydantic import BaseModel, Field
class Movies(BaseModel):
    title:str = Field(description="Movie Title")
    year:int = Field(description="Movie Released Year.")

class AllMovies(BaseModel):
    movies:List[Movies]



movie_llm = llm.with_structured_output(Movies)
```
