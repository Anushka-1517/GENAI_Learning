```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b")
out = StrOutputParser()
```

```python
### My Self Chain Node

def transform_case(result:str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a translator and translate input in {language}"),
    ("user", "{query}")
])

```

```python
#final_prompts = prompts.format_messages(language = "Hinglish", query = "how are you")
#res = llm.invoke(final_prompts)
#print(res.content)
```

```python
### Chains - Runnable
```

```python
chains = prompts | llm | out | transform_case
```

```python
res = chains.invoke({"language":"Hinglish", "query":"I love you"})
print(res)
```
