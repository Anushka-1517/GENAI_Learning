```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

```python
prompts =[
    {"role":"user", "content":"Hello, My name is Anushka Sangal."},
    {"role":"ai", "content":"Hello Anushka! Nice to meet you. How can I assist you today?"},
    {"role":"user","content":"What is my name?"}

]
res=llm.invoke(prompts)
```

```python
res.content
```

```python
### Manage History
```

```python
from langchain_core.messages import HumanMessage, AIMessage

history = []

# Message 1
history.append(HumanMessage(content="My name is Anushka."))

res = llm.invoke(history)
print("AI:", res.content)

history.append(AIMessage(content=res.content))

# Message 2
history.append(HumanMessage(content="What is my name?"))

res = llm.invoke(history)
print("AI:", res.content)

history.append(AIMessage(content=res.content))
```
